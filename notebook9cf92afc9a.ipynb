```python
# Ô CODE ĐẦU TIÊN - TẢI DỮ LIỆU
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Đọc dữ liệu Titanic
train_df = pd.read_csv('/kaggle/input/titanic/train.csv')

print("✅ ĐÃ TẢI DỮ LIỆU THÀNH CÔNG!")
print("Kích thước dữ liệu:", train_df.shape)
print("\n5 dòng đầu tiên:")
train_df.head()
```

```python
# Ô CODE THỨ HAI - KIỂM TRA CHẤT LƯỢNG
print("=== THÔNG TIN TỔNG QUAN ===")
train_df.info()

print("\n=== DỮ LIỆU THIẾU ===")
missing_data = train_df.isnull().sum()
print(missing_data[missing_data > 0])  # Chỉ hiển thị cột có dữ liệu thiếu

print("\n=== THỐNG KÊ CƠ BẢN ===")
train_df.describe()
```

```python
# Ô CODE THỨ BA - PHÂN TÍCH KHÁM PHÁ
print("=== PHÂN TÍCH TỶ LỆ SỐNG SÓT ===")

# Tạo subplot cho nhiều biểu đồ
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Biểu đồ 1: Giới tính vs Sống sót
sns.barplot(x='Sex', y='Survived', data=train_df, ax=axes[0,0], palette='pastel')
axes[0,0].set_title('TỶ LỆ SỐNG SÓT THEO GIỚI TÍNH')
axes[0,0].set_ylabel('Tỷ lệ sống sót')

# Biểu đồ 2: Hạng vé vs Sống sót
sns.barplot(x='Pclass', y='Survived', data=train_df, ax=axes[0,1], palette='Set2')
axes[0,1].set_title('TỶ LỆ SỐNG SÓT THEO HẠNG VÉ')
axes[0,1].set_ylabel('Tỷ lệ sống sót')

# Biểu đồ 3: Cảng lên tàu vs Sống sót
sns.barplot(x='Embarked', y='Survived', data=train_df, ax=axes[1,0], palette='coolwarm')
axes[1,0].set_title('TỶ LỆ SỐNG SÓT THEO CẢNG')
axes[1,0].set_ylabel('Tỷ lệ sống sót')

# Biểu đồ 4: Số người thân vs Sống sót
train_df['FamilySize'] = train_df['SibSp'] + train_df['Parch'] + 1
sns.barplot(x='FamilySize', y='Survived', data=train_df, ax=axes[1,1], palette='viridis')
axes[1,1].set_title('TỶ LỆ SỐNG SÓT THEO KÍCH THƯỚC GIA ĐÌNH')
axes[1,1].set_ylabel('Tỷ lệ sống sót')

plt.tight_layout()
plt.show()

# In tỷ lệ phần trăm cụ thể
print("\n=== TỶ LỆ SỐNG SÓT CHI TIẾT ===")
print("Theo giới tính:")
print(train_df.groupby('Sex')['Survived'].mean().round(3) * 100)
print("\nTheo hạng vé:")
print(train_df.groupby('Pclass')['Survived'].mean().round(3) * 100)
```

```python
# Ô CODE 2.1 - XỬ LÝ GIÁ TRỊ THIẾU
print("=== XỬ LÝ DỮ LIỆU THIẾU ===")

# Kiểm tra dữ liệu thiếu trước khi xử lý
print("Dữ liệu thiếu TRƯỚC KHI xử lý:")
print(train_df.isnull().sum())

# Xử lý cột Age - thay thế bằng giá trị trung vị
median_age = train_df['Age'].median()
train_df['Age'].fillna(median_age, inplace=True)
print(f"\n✅ Đã thay thế {train_df['Age'].isnull().sum()} giá trị thiếu trong Age bằng median: {median_age}")

# Xử lý cột Embarked - thay thế bằng giá trị phổ biến nhất
most_common_embarked = train_df['Embarked'].mode()[0]
train_df['Embarked'].fillna(most_common_embarked, inplace=True)
print(f"✅ Đã thay thế {train_df['Embarked'].isnull().sum()} giá trị thiếu trong Embarked bằng: '{most_common_embarked}'")

# Xử lý cột Cabin - tạo biến mới "Has_Cabin"
train_df['Has_Cabin'] = train_df['Cabin'].notnull().astype(int)
print(f"✅ Đã tạo biến Has_Cabin: {train_df['Has_Cabin'].sum()} hành khách có thông tin Cabin")

# Kiểm tra lại sau khi xử lý
print("\nDữ liệu thiếu SAU KHI xử lý:")
print(train_df.isnull().sum())

# Xem kết quả
print("\n=== THỐNG KÊ AGE SAU KHI XỬ LÝ ===")
print(train_df['Age'].describe())
```

```python
# Ô CODE 2.2 - MÃ HÓA BIẾN
print("=== MÃ HÓA BIẾN PHÂN LOẠI ===")

# Mã hóa giới tính thành số
train_df['Sex_Code'] = train_df['Sex'].map({'male': 0, 'female': 1})
print("✅ Đã mã hóa Sex:")
print(train_df[['Sex', 'Sex_Code']].head())

# Mã hóa cảng lên tàu
train_df['Embarked_Code'] = train_df['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})
print(f"\n✅ Đã mã hóa Embarked:")
print(train_df[['Embarked', 'Embarked_Code']].value_counts())

# Mã hóa hạng vé (giữ nguyên vì đã là số)
print(f"\n✅ Pclass đã là số, không cần mã hóa")

# Kiểm tra các cột đã mã hóa
print("\n=== CÁC CỘT SỐ SAU KHI MÃ HÓA ===")
print(train_df[['Sex_Code', 'Embarked_Code', 'Pclass']].head())
```

```python
# Ô CODE 2.3 - FEATURE ENGINEERING
print("=== TẠO BIẾN MỚI ===")

# 1. Kích thước gia đình
train_df['FamilySize'] = train_df['SibSp'] + train_df['Parch'] + 1
print("✅ Đã tạo FamilySize:")
print(train_df['FamilySize'].value_counts().sort_index())

# 2. Đi một mình hay không
train_df['IsAlone'] = (train_df['FamilySize'] == 1).astype(int)
print(f"\n✅ Đã tạo IsAlone: {train_df['IsAlone'].sum()} người đi một mình")

# 3. Nhóm tuổi
train_df['Age_Group'] = pd.cut(train_df['Age'], 
                              bins=[0, 12, 18, 35, 60, 100], 
                              labels=['Child', 'Teen', 'Adult', 'Middle', 'Senior'])
print("\n✅ Đã tạo Age_Group:")
print(train_df['Age_Group'].value_counts())

# 4. Title từ tên
train_df['Title'] = train_df['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)
print(f"\n✅ Đã trích xuất Title:")
print(train_df['Title'].value_counts())

# Hiển thị kết quả cuối cùng
print("\n=== DỮ LIỆU SAU KHI XỬ LÝ HOÀN THÀNH ===")
print("Các cột mới:", [col for col in train_df.columns if col not in ['PassengerId', 'Name', 'Ticket', 'Cabin']])
print(f"\nKích thước dữ liệu: {train_df.shape}")
print("\n5 dòng đầu tiên của dữ liệu đã xử lý:")
train_df[['Survived', 'Pclass', 'Sex_Code', 'Age', 'FamilySize', 'IsAlone', 'Age_Group', 'Title']].head()
```

```python
# Ô CODE 3.1 - CHUẨN BỊ MÔ HÌNH
print("=== CHUẨN BỊ DỮ LIỆU CHO MÔ HÌNH ===")

# Import các thư viện machine learning
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score, train_test_split

# Chọn các features quan trọng
features = ['Pclass', 'Sex_Code', 'Age', 'Fare', 'FamilySize', 'Embarked_Code']
X = train_df[features]
y = train_df['Survived']

print("✅ ĐÃ CHUẨN BỊ DỮ LIỆU:")
print(f"- Số features: {len(features)}")
print(f"- Features sử dụng: {features}")
print(f"- Kích thước X: {X.shape}")
print(f"- Kích thước y: {y.shape}")
print(f"- Số người sống sót: {y.sum()} / {len(y)} ({y.mean()*100:.1f}%)")

# Hiển thị dữ liệu đầu vào
print("\n=== DỮ LIỆU ĐẦU VÀO (5 dòng đầu) ===")
X.head()
```

```python
# Ô CODE 3.2 - ĐÁNH GIÁ MÔ HÌNH
print("=== SO SÁNH CÁC MÔ HÌNH ===")

# Danh sách các mô hình sẽ thử
models = {
    'Logistic Regression': LogisticRegression(),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),
    'Decision Tree': DecisionTreeClassifier(random_state=42)
}

# Đánh giá từng mô hình bằng cross-validation
results = {}
for name, model in models.items():
    cv_scores = cross_val_score(model, X, y, cv=5, scoring='accuracy')
    results[name] = cv_scores
    print(f"{name}:")
    print(f"  - Độ chính xác: {cv_scores.mean():.3f} (+/- {cv_scores.std() * 2:.3f})")
    print(f"  - Chi tiết: {cv_scores}")

# Tìm mô hình tốt nhất
best_model_name = max(results, key=lambda x: results[x].mean())
print(f"\n🎯 MÔ HÌNH TỐT NHẤT: {best_model_name}")
print(f"   Độ chính xác: {results[best_model_name].mean():.3f}")

# Vẽ biểu đồ so sánh
plt.figure(figsize=(10, 6))
box_data = [results[name] for name in models.keys()]
plt.boxplot(box_data, labels=models.keys())
plt.title('SO SÁNH ĐỘ CHÍNH XÁC CỦA CÁC MÔ HÌNH')
plt.ylabel('Độ chính xác')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Ô CODE 3.3 - HUẤN LUYỆN MÔ HÌNH CUỐI CÙNG
print("=== HUẤN LUYỆN MÔ HÌNH TỐT NHẤT ===")

# Chọn mô hình tốt nhất từ kết quả trên
if best_model_name == 'Logistic Regression':
    final_model = LogisticRegression()
elif best_model_name == 'Random Forest':
    final_model = RandomForestClassifier(n_estimators=100, random_state=42)
else:
    final_model = DecisionTreeClassifier(random_state=42)

# Huấn luyện trên toàn bộ dữ liệu
final_model.fit(X, y)
print(f"✅ ĐÃ HUẤN LUYỆN MÔ HÌNH: {best_model_name}")

# Kiểm tra độ chính xác trên chính dữ liệu huấn luyện
train_accuracy = final_model.score(X, y)
print(f"✅ Độ chính xác trên dữ liệu huấn luyện: {train_accuracy:.3f}")

# Phân tích importance của các features (nếu là Random Forest)
if best_model_name == 'Random Forest':
    print(f"\n=== QUAN TRỌNG CỦA CÁC FEATURE ===")
    feature_importance = pd.DataFrame({
        'feature': features,
        'importance': final_model.feature_importances_
    }).sort_values('importance', ascending=False)
    
    print(feature_importance)
    
    # Vẽ biểu đồ importance
    plt.figure(figsize=(10, 6))
    sns.barplot(data=feature_importance, x='importance', y='feature')
    plt.title('MỨC ĐỘ QUAN TRỌNG CỦA CÁC ĐẶC TRƯNG')
    plt.xlabel('Mức độ quan trọng')
    plt.show()

print("\n🎉 MÔ HÌNH ĐÃ SẴN SÀNG ĐỂ DỰ ĐOÁN!")
```

```python
# Ô CODE 4.1 - XỬ LÝ DỮ LIỆU TEST
print("=== XỬ LÝ DỮ LIỆU TEST ===")

# Kiểm tra dữ liệu test ban đầu
print("Dữ liệu test ban đầu:")
print(f"Kích thước: {test_df.shape}")
print(test_df.info())

# ÁP DỤNG CÁC BƯỚC XỬ LÝ GIỐNG NHƯ TRAIN DATA

# 1. Xử lý giá trị thiếu
print("\n=== XỬ LÝ GIÁ TRỊ THIẾU ===")

# Age - dùng cùng median_age từ train data
test_df['Age'].fillna(median_age, inplace=True)
print(f"✅ Đã xử lý Age thiếu")

# Fare - nếu có thiếu, dùng median từ train
if test_df['Fare'].isnull().sum() > 0:
    fare_median = train_df['Fare'].median()
    test_df['Fare'].fillna(fare_median, inplace=True)
    print(f"✅ Đã xử lý Fare thiếu")

# Embarked - dùng cùng most_common_embarked từ train data
test_df['Embarked'].fillna(most_common_embarked, inplace=True)
print(f"✅ Đã xử lý Embarked thiếu")

# 2. Mã hóa biến (giống hệt train data)
test_df['Sex_Code'] = test_df['Sex'].map({'male': 0, 'female': 1})
test_df['Embarked_Code'] = test_df['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})

# 3. Feature engineering (giống hệt train data)
test_df['FamilySize'] = test_df['SibSp'] + test_df['Parch'] + 1
test_df['IsAlone'] = (test_df['FamilySize'] == 1).astype(int)

print("\n✅ ĐÃ XỬ LÝ XONG DỮ LIỆU TEST")
print(f"Kích thước sau xử lý: {test_df.shape}")
print(f"Dữ liệu thiếu còn lại: {test_df[features].isnull().sum().sum()}")

# Kiểm tra dữ liệu test đã xử lý
print("\n=== DỮ LIỆU TEST ĐÃ XỬ LÝ (5 dòng đầu) ===")
test_df[features].head()
```

```python
# Ô CODE 4.2 - DỰ ĐOÁN
print("=== TẠO DỰ ĐOÁN ===")

# Đảm bảo test data có đủ các features cần thiết
print("Features sử dụng cho dự đoán:")
print(features)

# Kiểm tra xem test data có đủ tất cả features không
missing_features = [f for f in features if f not in test_df.columns]
if missing_features:
    print(f"❌ THIẾU FEATURES: {missing_features}")
else:
    print("✅ Đủ tất cả features")

# Tạo dự đoán
try:
    predictions = final_model.predict(test_df[features])
    print(f"✅ ĐÃ TẠO DỰ ĐOÁN CHO {len(predictions)} HÀNH KHÁCH")
    
    # Thống kê kết quả dự đoán
    unique, counts = np.unique(predictions, return_counts=True)
    prediction_stats = dict(zip(unique, counts))
    
    print(f"\n=== THỐNG KÊ DỰ ĐOÁN ===")
    print(f"Số người được dự đoán KHÔNG sống sót (0): {prediction_stats.get(0, 0)}")
    print(f"Số người được dự đoán SỐNG SÓT (1): {prediction_stats.get(1, 0)}")
    print(f"Tỷ lệ sống sót dự đoán: {prediction_stats.get(1, 0)/len(predictions)*100:.1f}%")
    
    # Hiển thị 10 dự đoán đầu tiên
    print(f"\n10 DỰ ĐOÁN ĐẦU TIÊN:")
    for i, (passenger_id, pred) in enumerate(zip(test_df['PassengerId'].head(10), predictions[:10])):
        status = "SỐNG SÓT" if pred == 1 else "KHÔNG sống sót"
        print(f"  Passenger {passenger_id}: {status}")
        
except Exception as e:
    print(f"❌ LỖI KHI DỰ ĐOÁN: {e}")
```

```python
# Ô CODE 4.3 - TẠO FILE SUBMISSION
print("=== TẠO FILE NỘP BÀI ===")

# Tạo DataFrame submission
submission = pd.DataFrame({
    'PassengerId': test_df['PassengerId'],
    'Survived': predictions
})

# Kiểm tra file submission
print("✅ ĐÃ TẠO FILE SUBMISSION:")
print(f"Kích thước: {submission.shape}")
print(f"\n5 dòng đầu:")
print(submission.head())

print(f"\n5 dòng cuối:")
print(submission.tail())

# Lưu file CSV
submission.to_csv('submission.csv', index=False)
print(f"\n💾 ĐÃ LƯU FILE: submission.csv")

# Thông báo quan trọng
print("\n" + "="*50)
print("🎉 CHÚC MỪNG! BẠN ĐÃ HOÀN THÀNH DỰ ÁN!")
print("="*50)
print("\n📍 CÁCH NỘP BÀI LÊN KAGGLE:")
print("1. Ở panel bên phải Kaggle, click 'Add data'")
print("2. Chọn 'Your Work' → Upload file 'submission.csv'")
print("3. Vào trang Titanic Competition → 'Submit Predictions'")
print("4. Chọn file vừa upload → 'Make Submission'")
print("5. Xem điểm số của bạn trên Leaderboard!")

print(f"\n📊 DỰ KIẾN ĐIỂM: Khoảng {results[best_model_name].mean()*100:.1f}%")
print("   (Độ chính xác có thể thay đổi khi nộp bài thực tế)")

# Hiển thị file submission cuối cùng
print(f"\n=== FILE SUBMISSION.CSV ===")
print(submission.head(10))
```

```python
# Ô CODE KIỂM TRA BỔ SUNG
print("=== ĐÁNH GIÁ CHI TIẾT MÔ HÌNH ===")

# 1. Độ chính xác tổng thể
cv_scores = results[best_model_name]
print(f"🎯 ĐỘ CHÍNH XÁC DỰ KIẾN: {cv_scores.mean()*100:.2f}%")

# 2. Phân tích dự đoán chi tiết
print(f"\n📈 PHÂN BỐ DỰ ĐOÁN:")
survived_count = np.sum(predictions)
not_survived_count = len(predictions) - survived_count
print(f"- Dự đoán KHÔNG sống sót: {not_survived_count} hành khách ({not_survived_count/len(predictions)*100:.1f}%)")
print(f"- Dự đoán SỐNG SÓT: {survived_count} hành khách ({survived_count/len(predictions)*100:.1f}%)")

# 3. So sánh với tập train
print(f"\n📊 SO SÁNH VỚI TẬP TRAIN:")
train_survival_rate = train_df['Survived'].mean() * 100
test_survival_rate = (survived_count / len(predictions)) * 100
print(f"- Tỷ lệ sống sót thực tế (train): {train_survival_rate:.1f}%")
print(f"- Tỷ lệ sống sót dự đoán (test): {test_survival_rate:.1f}%")

# 4. Hiển thị một số dự đoán mẫu
print(f"\n🔍 10 DỰ ĐOÁN MẪU:")
sample_predictions = list(zip(test_df['PassengerId'].head(10), predictions[:10]))
for passenger_id, pred in sample_predictions:
    status = "SỐNG SÓT 🟢" if pred == 1 else "KHÔNG sống sót 🔴"
    print(f"   Hành khách {passenger_id}: {status}")
```
